# Prepare the forecasting dataset

## Demand Prediction

```python
# Select the total freight movement series

target_column = "Total (million net tonne kilometers)"

forecast_df = orr_1314[
    ["Time period", target_column]
].copy()

# Rename columns for easier modelling
forecast_df = forecast_df.rename(
    columns={
        "Time period": "period",
        target_column: "freight_demand"
    }
)

# Convert target to numeric
forecast_df["freight_demand"] = pd.to_numeric(
    forecast_df["freight_demand"],
    errors="coerce"
)

# Extract financial year and period number
extracted = forecast_df["period"].str.extract(
    r"Apr (\d{4}) to Mar (\d{4}) \(Period (\d{2})\)"
)

forecast_df["financial_year_start"] = extracted[0].astype(int)
forecast_df["period_number"] = extracted[2].astype(int)

# Ensure chronological order
forecast_df = forecast_df.sort_values(
    ["financial_year_start", "period_number"]
).reset_index(drop=True)

print("Forecasting dataset:")
display(forecast_df.head())

print("\nLast observations:")
display(forecast_df.tail())

print("\nDataset shape:", forecast_df.shape)
print("\nMissing values:")
print(forecast_df.isnull().sum())
```

# Create forecasting features

```python
# Create lag features

forecast_df["lag_1"] = forecast_df["freight_demand"].shift(1)
forecast_df["lag_2"] = forecast_df["freight_demand"].shift(2)
forecast_df["lag_3"] = forecast_df["freight_demand"].shift(3)
forecast_df["lag_4"] = forecast_df["freight_demand"].shift(4)
forecast_df["lag_13"] = forecast_df["freight_demand"].shift(13)

# Previous four-period rolling average
forecast_df["rolling_mean_4"] = (
    forecast_df["freight_demand"]
    .shift(1)
    .rolling(window=4)
    .mean()
)

display(forecast_df.tail(15))
```

# Formulate the mathematical optimisation model

```python
# Define a reusable optimisation function.
# The unmet-demand penalty should be sufficiently high to prioritise serving demand.
def optimise_freight(demand, capacities, costs, unmet_penalty=1000.0, solver_msg=False):
    if demand < 0:
        raise ValueError("Demand must be non-negative.")
    if set(capacities) != set(costs):
        raise ValueError("Capacity and cost dictionaries must contain the same routes.")
    if any(v < 0 for v in capacities.values()):
        raise ValueError("Route capacities must be non-negative.")
    if any(v < 0 for v in costs.values()):
        raise ValueError("Transport costs must be non-negative.")

    route_names = list(capacities.keys())
    problem = pulp.LpProblem("Railway_Freight_Allocation", pulp.LpMinimize)

    allocation_vars = pulp.LpVariable.dicts(
        "Freight_Allocation", route_names, lowBound=0, cat="Continuous"
    )
    unmet_var = pulp.LpVariable("Unmet_Demand", lowBound=0, cat="Continuous")

    problem += (
        pulp.lpSum(costs[r] * allocation_vars[r] for r in route_names)
        + unmet_penalty * unmet_var
    )

    for r in route_names:
        problem += allocation_vars[r] <= capacities[r], f"Capacity_{r}"

    problem += (
        pulp.lpSum(allocation_vars[r] for r in route_names) + unmet_var == demand,
        "Demand_Balance"
    )

    solver = pulp.PULP_CBC_CMD(msg=solver_msg)
    problem.solve(solver)

    status = pulp.LpStatus[problem.status]
    if status != "Optimal":
        raise RuntimeError(f"Optimisation did not find an optimal solution: {status}")

    allocated_by_route = {r: float(pulp.value(allocation_vars[r]) or 0.0) for r in route_names}
    unmet_value = float(pulp.value(unmet_var) or 0.0)
    total_allocated = sum(allocated_by_route.values())

    result = pd.DataFrame({
        "Route": route_names,
        "Allocated Freight": [allocated_by_route[r] for r in route_names],
        "Capacity": [capacities[r] for r in route_names],
        "Cost per Unit": [costs[r] for r in route_names]
    })
    return {
        "status": status,
        "allocation": allocated_by_route,
        "unmet_demand": unmet_value,
        "total_allocated": total_allocated,
        "total_cost": float(pulp.value(problem.objective)),
        "details": result,
        "model": problem
    }
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Prepare forecasting dataset
target_column = 'Total (million net tonne kilometers)'
forecast_df = orr_1314[['Time period', target_column]].copy()
forecast_df = forecast_df.rename(columns={'Time period': 'period', target_column: 'freight_demand'})
forecast_df['freight_demand'] = pd.to_numeric(forecast_df['freight_demand'], errors='coerce')

extracted = forecast_df['period'].str.extract(r'Apr (\d{4}) to Mar (\d{4}) \(Period (\d{2})\)')
forecast_df['financial_year_start'] = extracted[0].astype(int)
forecast_df['period_number'] = extracted[2].astype(int)
forecast_df = forecast_df.sort_values(['financial_year_start', 'period_number']).reset_index(drop=True)

# 2. Recreate features
forecast_df['lag_1'] = forecast_df['freight_demand'].shift(1)
forecast_df['lag_2'] = forecast_df['freight_demand'].shift(2)
forecast_df['lag_3'] = forecast_df['freight_demand'].shift(3)
forecast_df['lag_4'] = forecast_df['freight_demand'].shift(4)
forecast_df['lag_13'] = forecast_df['freight_demand'].shift(13)
forecast_df['rolling_mean_4'] = forecast_df['freight_demand'].shift(1).rolling(window=4).mean()

model_df = forecast_df.dropna().copy()
feature_columns = ['lag_1', 'lag_2', 'lag_3', 'lag_4', 'lag_13', 'rolling_mean_4']

X = model_df[feature_columns]
y = model_df['freight_demand']
split_index = int(len(model_df) * 0.80)

X_train, X_test = X.iloc[:split_index], X.iloc[split_index:]
y_train, y_test = y.iloc[:split_index], y.iloc[split_index:]

# 3. Train models and generate predictions
rf_model = RandomForestRegressor(n_estimators=200, max_depth=3, min_samples_leaf=2, random_state=42)
rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

xgb_model = XGBRegressor(n_estimators=100, max_depth=2, learning_rate=0.03, subsample=0.8, colsample_bytree=0.8, reg_lambda=5, objective='reg:squarederror', random_state=42)
xgb_model.fit(X_train, y_train)
xgb_predictions = xgb_model.predict(X_test)

baseline_predictions = model_df['lag_1'].iloc[split_index:].values

# 4. Generate results DataFrame
def calculate_metrics(actual, predicted):
    mae = mean_absolute_error(actual, predicted)
    rmse = np.sqrt(mean_squared_error(actual, predicted))
    wape = (np.sum(np.abs(actual - predicted)) / np.sum(np.abs(actual))) * 100
    return mae, rmse, wape

results = []
models = {
    'Naive Baseline': baseline_predictions,
    'Random Forest': rf_predictions,
    'XGBoost': xgb_predictions
}

for model_name, predictions in models.items():
    mae, rmse, wape = calculate_metrics(y_test.values, predictions)
    results.append({'Model': model_name, 'MAE': mae, 'RMSE': rmse, 'WAPE (%)': wape})

results_df = pd.DataFrame(results)
print('Variables y_test, rf_predictions, and results_df successfully restored!')
display(results_df.round(3))
```
